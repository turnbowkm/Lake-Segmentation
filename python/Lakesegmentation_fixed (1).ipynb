```python
import geopandas as gpd

# 1. Load your GeoDataFrame
gdf = gpd.read_file(filepath)

# 2. Reproject to a local projected CRS (Example: UTM Zone 15N - EPSG:32615)
# Replace 'EPSG:32615' with the appropriate UTM zone or local projected CRS for your geographic area
gdf_projected = gdf.to_crs(epsg=32615)

# 3. Calculate the perimeter in meters (standard for most projected CRS)
perimeter_meters = gdf_projected.geometry.length

# 4. Convert meters to yards (1 meter = 1.09361 yards)
gdf['perimtr_yd'] = perimeter_meters * 1.09361

# View the result
print(gdf[['perimtr_yd']])
```

```python
bins = [0, 10, 50, 100, 200, 300, 400, 500, 800, np.inf] # np.inf for the upper bound of the last category
labels = lake_size_lookup_df['size'].tolist() # Use the 'size' strings as labels
gdf['Lake_Size_Category'] = pd.cut(
    gdf['WATER_SIZE'],
    bins=bins,
    labels=labels,
    right=False, # Important for correct binning based on your ranges
    include_lowest=True # Include 0 in the first bin
)
gdf = gdf.merge(
    lake_size_lookup_df,
    left_on='Lake_Size_Category',
    right_on='size',
    how='left' # Use left join to keep all lakes from your GeoDataFrame
)
gdf['dist'] = (gdf['perimtr_yd']* 0.9144) / gdf['ttlptts']
#gdf.drop('size', axis=1, inplace=True)
gdf
```

```python
all_points = []

for i in range(len(gdf)):
    # Reproject this lake's geometry to UTM (meters) so distances match `dist` (meters)
    geom_utm = gpd.GeoSeries([gdf.geometry.iloc[i]], crs=gdf.crs).to_crs(epsg=32615).iloc[0]
    boundary_line = geom_utm.boundary
    dis = gdf['dist'].iloc[i]

    if pd.isna(dis) or dis <= 0:
        print(f"Skipping index {i} ({gdf['ASSESSMENT'].iloc[i]}): invalid dist ({dis})")
        continue

    fixed_interval = dis
    distances = np.arange(0, boundary_line.length, fixed_interval)
    points_utm = [boundary_line.interpolate(distance) for distance in distances]

    lake_points_gdf = gpd.GeoDataFrame(geometry=points_utm, crs='EPSG:32615')
    lake_points_gdf = lake_points_gdf.to_crs(gdf.crs)  # back to lat/lon
    lake_points_gdf = lake_points_gdf.reset_index(level=0)
    lake_points_gdf['index'] = lake_points_gdf['index'] + 1
    lake_points_gdf['indexa'] = lake_points_gdf['index'].apply(lambda x: 'S' if x % 3 else 'D')

    name = gdf['ASSESSMENT'].iloc[i]
    lake_points_gdf['code'] = name + lake_points_gdf['index'].astype('string') + '_Macroph-' + lake_points_gdf['indexa']
    lake_points_gdf = lake_points_gdf.drop(['index', 'indexa'], axis=1)

    # Per-lake shapefile (same as before, now for every lake instead of just index 0)
    lake_points_gdf.to_file('C:/Users/kevin.turnbow/GIS/' + name + '_processed.shp')

    all_points.append(lake_points_gdf)

# Combine every lake's points into one GeoDataFrame for the GPX export below
sampled_points_gdf = pd.concat(all_points, ignore_index=True)
sampled_points_gdf = gpd.GeoDataFrame(sampled_points_gdf, geometry='geometry', crs=gdf.crs)

print(f"Generated {len(sampled_points_gdf)} points across {len(all_points)} lakes")
sampled_points_gdf.explore(color='blue')

sampled_points_gdf.explore(m=m, color="red")
```

```python
sampled_points_gdf = sampled_points_gdf.set_geometry(sampled_points_gdf.geometry.force_2d())
sampled_points_gdf
```

```python
# GPX only supports a fixed waypoint schema (name, desc, cmt, ele, time, etc.) -
# it will reject a custom field like 'code' unless renamed to a supported field.
gpx_export_gdf = sampled_points_gdf.rename(columns={'code': 'name'})[['name', 'geometry']]

gpx_export_gdf.to_file('C:/Users/kevin.turnbow/GIS/LAKES25_processed.gpx', driver="GPX")
```
